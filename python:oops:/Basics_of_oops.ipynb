```python
class listop:
    l = [1,2,3,4,5,6]
    def extract_even(self,l):
        l1 = []
        for i in l:
            if i % 2 == 0:
                l1.append(i)
        return l1
    
    def extract_odd(self,l):
        l1 = []
        for i in l:
            if i % 2 != 0:
                l1.append(i)
        return l1
```

```python
class listop:
    
    def __init__(self,l):
        self.l = l

    def extract_even(self):
        l = self.l
        l1 = []
        for i in l:
            if i % 2 == 0:
                l1.append(i)
        return l1
    
    def extract_odd(self):
        l = self.l
        l1 = []
        for i in l:
            if i % 2 != 0:
                l1.append(i)
        return l1
```

```python
obj1 = listop([1,2,3,4,5,6])
```

```python
obj1.extract_even()
```

```python
obj2 = listop([1,23,34,5645,65,76,89,88])
obj2.extract_even()
```
