```python
#factorial
def fact_num(n):
    if n==0:
        return 1
    else:
        return n*fact_num(n-1)
```

```python
fact_num(5)
```

```python
#factorial by lembda ex.

fact_num= lambda n: 1 if n == 0 else n*fact_num(n-1)
```

```python
fact_num(5)
```

```python
# map >> executes a specified function for each of item of an iterable
```

```python
help(map)
```

```python
l = [1,2,3,4,5]

def square(l):
    sq = []
    for i in l:
        sq.append(i**2)
    return sq
```

```python
square(l)
```

```python
# reduce >> folding/reduction
```

```python
from functools import reduce
```

```python
number = [1,2,3,43,223,100,334]
reduce(lambda x,y: x if  x>y else y , number)
```

```python
l = [2,3,1,4,22,3]
reduce(lambda x,y : x+y ,l)
```

```python
words = ["Data","science","Course"]
reduce(lambda x,y : x+" "+y, words)
```

```python
def fact(n):
    return reduce(lambda x,y: x*y ,range(1,n+1))
```

```python
fact(5)
```

```python
def check_even(num):
    if num % 2 == 0:
        return True
    return False
```

```python
num = [23,34,32,54,56]
list(filter(check_even,num))
```
