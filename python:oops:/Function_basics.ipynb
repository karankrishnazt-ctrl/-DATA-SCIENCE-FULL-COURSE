```python
def only_numeric(a):
    n = []
    for i in a:
        if type(i) == int or type(i) == float:
            n.append(i)

    return n
```

```python
only_numeric([1,2,3,4.3,5,3+4j,"lalo","raghu",True])
```

```python
def only_numeric(a):
    n = []
    for i in a:
        if type(i) == list:
            for j in i:
                if type(j) == int or type(j) == float:
                    n.append(j)
        else: 
            if type(i) == int or type(i) == float:
                n.append(i)

    return n
```

```python
only_numeric([1,2,3,4,"kk","sk",[3,6,7]])
```

```python
#variable length argument

def sum_(*args):
    s = 0
    for i in args:
        s = s+i
    return s
```

```python
def return_list(*args):
    l = []
    for i in args:
        if type(i) == list:
            l.append(i)
    return l
```

```python
def test(**kwargs):
    for i in kwargs.keys():
        if type(kwargs[i]) == list:
            return i, kwargs[i]
```

```python
test(a = 2, b = 3, c = 32,d=[12,21,41,31],e=(32,42,23),f=[54,64,34])
```

```python
#function inside function 
def marks_in_subjects(**kwargs):
    def total_marks(marks_list):
        return sum(marks_list)
    marks_list = []
    for subject,marks in kwargs.items():
        marks_list.append(marks)
    return total_marks(marks_list)
```

```python
marks_in_subjects(a=21,b=39,c=30)
```
