```python
n = 7
i = 1

while i < n:
    print(i)
    i = i + 1
    if i == 3:
        break
else:
    print("this will execute when while loop is completed")
```

```python
#continue -> to skip that iterator
n = 7
i = 1

while i < n:
    i = i + 1
    if i == 3:
        continue
    print(i)
else:
    print("this will execute when while loop is completed")
```
