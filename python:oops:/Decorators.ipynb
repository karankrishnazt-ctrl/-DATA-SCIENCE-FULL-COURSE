```python
# decorators >> allows to modify/extend the existing behaviour of function or class without modifying it 
```

```python
def My_decorator_func():
    print("The lines being printed before the comp. ")
    print(1+11)
    print("The lines being printed after the comp. ")
```

```python
My_decorator_func()
```

```python
def My_decorator_func(func): # decorator func that takes another func as arguement
    def wrapper(): # add the funtionality before and after calling function
        print("The lines being printed before the comp. ")
        func()
        print("The lines being printed after the comp. ")
    return wrapper
```

```python
@My_decorator_func
def say_hello():
    print ("hello")
# when say_hello() is called , it actually calling wrapper() with in turn call say_hello()
```

```python
say_hello()
```

```python
import time
def time_decorator(func):
    def timer():
        start = time.time()
        func()
        end = time.time()
        print(end-start)
    return timer
```

```python
@time_decorator
def func_test():
    print(1100**2)
```

```python
func_test()
```

```python
class MyDecorator:
    def __init__(self,func):
        self.func = func

    def __call__(self):  
        print("The lines being printed before the comp. ")
        self.func()
        print("The lines being printed after the comp. ")  
```

```python
@MyDecorator # __call__  is a special method which is invoked when you call a decorator clas instance
def say_hello():
    print("hello")
```

```python
say_hello()
```

```python
#static method >> which can be called without creating an instance of class
```

```python
class Math:

    def add(self,x,y):
        return x+y
```

```python
a = Math() # make object/instance
```

```python
a.add(2,5) # this was regular class 
```

```python
class Math:

    @staticmethod
    def add(x,y):
        return x+y
```

```python
Math.add(4.5,4.6)
```

```python
# class method >> takes class itself as first argument

class Math:
    @classmethod #takes referene to the class itself to modify and access class level attributes
    def add(cls,x,y):
        return cls.__name__, x+y # cls.__name__ >> Math
```

```python
# prooperty Decorator >> it allows methods to be accessed as attribute

class Circle:
    def __init__(self,radius):
        self.radius = radius
    @property
    def area(self):
        radius = self.radius
        return 3.14 * radius ** 2
```

```python
c = Circle(3)
```

```python
c.area
```

```python
c.radius
```
