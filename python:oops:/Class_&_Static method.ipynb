```python
class Student():
    def __init__(self,name): #method of an instance
        self.name = name
```

```python
stud = Student("kk")
```

```python
stud.name
```

```python
class Student():
    def __init__(self,name): 
        self.name = name

    @classmethod # bound to the class
    def student_details(cls,name):
        return cls(name)
```

```python
obj1 = Student.student_details("k.krishna")
```

```python
obj1.name
```

```python
# To make external class method to be part of the class
class Student():

    total_student = 0
    def __init__(self,name): #instance
        self.name = name
        Student.total_student = self.total_student + 1

    @classmethod # bound to the class
    def get_total_student(cls):
        return cls.total_student
```

```python
def course_details(cls, course_name):
    print("The details of the course is:", course_name)
```

```python
Student.course_details = classmethod(course_details) # adding the extra class method
```

```python
Student.course_details("Data science")
```

```python
# delet any class method
del Student.course_details 
```

```python
Student.course_details("DS")
```

```python
Student.get_total_student()
```

```python
#static method >> don't have access or modify class, work with the parameter that is passed
```

```python
class Calculator:
    @staticmethod
    def add(x,y):
        return x+y
    
    @staticmethod
    def subtract(x,y):
        return x-y
```

```python
Calculator.add(6,4)
```

```python
Calculator.subtract(66.5,56)
```
