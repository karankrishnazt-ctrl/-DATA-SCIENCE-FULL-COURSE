```python
class Teacher_lecture:
    def lec_info(self):
        print ("This is lec info with teacher perspective")
```

```python
class Student_lecture:
    def lec_info(self):
        print ("This is lec info with student perspective")
```

```python
def parcer(class_obj):
    for i in class_obj:
        i.lec_info()
```

```python
obj1 = Teacher_lecture()
obj2 = Student_lecture()
```

```python
class_obj = [obj1,obj2]
```

```python
parcer(class_obj)
```

```python
# method overloading >> student is taking different forms, the last method overwrite the previous ones

class student:
    def student(self):
        print("welcome")
    def student(self,name = ""):
        print("welcome",name)
    def student(self,name = "",course = ""):
        print("welcome",name,course)
```

```python
stud = student()
```

```python
stud.student()
```

```python
stud.student("verma")
```

```python
stud.student("verma","BBA")
```

```python
# private >> within the class only its accessible, use two __ to make private

class Student():
    def __init__(self,name,degree):
        self.name = name
        self.__degree = degree

    def show(self):
#  accessing  the public and private data member
        print("name",self.name,'degree',self.__degree)
```

```python

stud = Student("Ramesh","Graduation")
```

```python
stud.name
```

```python
stud.__degree # it make private
```

```python
stud._Student__degree
```

```python
stud.show()
```

```python
class Car:
    def __init__(self,year,make,speed,model):
        self.__year = year
        self.__make = make
        self.__speed = speed
        self.__model = model

    def set_speed(self,speed):
        self.__speed = 0 if speed < 0 else speed

    def get_speed(self):
        return self.__speed
```

```python
class Bank:

    def __init__(self,balance):
        self.__balance = balance

    def deposit(self,amount):
        self.__balance = self.__balance + amount

    def without(self,amount):
        if self.__balance >= amount:
            self.__balance = self.__balance - amount 
            return True
        else:
            return False
        
    def get__balance(self):
        return self.__balance
```

```python
acc = Bank(1000)
```

```python
acc.deposit(11000)
```

```python
acc.get__balance()
```

```python
acc.without(12000)
```

```python
acc.get__balance()
```

```python
acc.without(200) 
```

```python
# Protected >> within the class and its subclass protected member can be accessed, (_)

class College:
    def __init__(self):
        self._collage_name = "PWskill"

class Student(College):
    def __init__(self,name):
        self.name = name
        College.__init__(self)

    def show(self):
        print("name",self.name,"College",self._collage_name)
```

```python
stud = Student("kk")
```

```python
stud.show()
```

```python
coll = College()
```

```python
coll._collage_name # you can see the variable but can not change it 
```

```python
class College:
    def __init__(self):
        self._collage_name = "PWskill"

class Student(College):
    def __init__(self,name):
        self.name = name
        super().__init__()

    def show(self):
        print("name",self.name,"College",self._collage_name)
```

```python
stud = Student("kk")
stud.show()
```
