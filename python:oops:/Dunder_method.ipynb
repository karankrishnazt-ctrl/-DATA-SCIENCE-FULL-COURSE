```python
#repr

class Myclass:
    def __init__(self,x):
        self.x = x

    def __repr__(self):
        return f"Myclass({self.x})"
```

```python
obj = Myclass(2)
```

```python
repr(obj) # return the string representation of odject
```

```python
class Point:
    def __init__(self,x,y):
        self.x = x
        self.y = y

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y
```

```python
p1 = Point(2,3)
p2 = Point(2,5)
```

```python
print(p1 == p2)
```
