```python
import csv

with open("example_csv.csv","w") as f:
    w = csv.writer(f)
    for i in data:
        w.writerow(i) 
```

```python
# Buffered reading >> it allows the program to read the data in chunks
#Buffered writing
```

```python
import io
with open("test_buf.txt","wb") as f:
    file = io.BufferedWriter(f)
    file.write(b"0000000000\n")
    file.write(b"1111111111\n")
    file.write(b"2222222222\n")
    file.write(b"The Art of Racing in the Rain is a 2019 American drama film directed by Simon Curtis and written by Mark Bomback, based on the 2008 novel by author Garth Stein \n")
    file.flush()
```

```python
with open("test_buf.txt","rb") as f:
    file = io.BufferedReader(f)
    data = file.read(200) # 200 bytes
    print(data)
```
