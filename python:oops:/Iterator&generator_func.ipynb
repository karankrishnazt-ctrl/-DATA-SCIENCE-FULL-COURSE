```python
#generayor function 
#generator is a type of function which does not return a single value , instad return a iterator object
  
```

```python
def square_nos(n):
    result = []
    for i in range(n):
        result.append(i ** 2)
    return result
squre_nos(5)
```

```python
def square_number_generator(n):
    for i in range(n):
        yield i ** 2
```

```python
gen = square_number_generator(5)
```

```python
next(gen)
```

```python
list(gen)
```

```python
# fibonacci function

def fib(n):
    a = 0
    b = 1
    for i in range(n):
        yield a
        a, b = b, a+b
```

```python
f = fib(100000)
```

```python
next(f)
```

```python
next(f)
```
