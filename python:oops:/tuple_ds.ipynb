```python
#______________________set operation_______________________
#there are 4 types
#1. union
#2. intersection
#3. difference
#4. symmetric
```

```python
s1 = {"hiking","reading","coading"}
s2 = {"coading","photoggraphy","travelling"}
```

```python
# intersection >> &
s1 & s2
```

```python
# union >> |
s1 | s2
```

```python
# difference >> -
s1-s2
```

```python
# symmetric difference
s1 ^ s2
```
