```python
df1 = pd.DataFrame({
    'key1' : [1,2,3,4,5,6],
    'key2' : [4,5,6,7,8,9],
    'key3' : [7,6,5,4,3,2]
})
```

```python
df1
```

```python
df2 = pd.DataFrame({
    'key1' : [1,43,5,4,6,7],
    'key4' : [32,3,2,45,65,8],
    'key5' : [12,32,43,5,46,7]
})
```

```python
df2
```

```python
pd.merge(df1,df2, how = 'inner')
```

```python
pd.merge(df1,df2, how = 'left')
```

```python
pd.merge(df1,df2, how = 'right')
```

```python
pd.merge(df1,df2, how = 'outer') # all the keys from both the dataframe
```

```python
df1
```

```python
df2
```

```python
pd.merge(df1, df2, how = 'cross') # combination of df1 and df2
```

```python
pd.merge(df1, df2, how = 'left', left_on='key2', right_on='key4')
```
